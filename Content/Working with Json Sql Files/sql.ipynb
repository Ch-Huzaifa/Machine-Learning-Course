### Common SQL Queries

1. SELECT — Retrieve Data

SELECT * FROM country;  
SELECT Name, Population FROM country;
___

2. WHERE — Filter Data

SELECT * FROM country
WHERE Continent = 'Asia';  
SELECT * FROM country
WHERE Population > 100000000;
___

3. ORDER BY — Sort Data

SELECT Name, Population
FROM country
ORDER BY Population DESC;  
SELECT Name, Population
FROM country
ORDER BY Population ASC;
___

4. LIMIT — Limit the Number of Rows

SELECT * FROM country
LIMIT 10;  
SELECT Name, Population
FROM country
ORDER BY Population DESC
LIMIT 5;
___

5. DISTINCT — Get Unique Values

SELECT DISTINCT Continent
FROM country;  
SELECT DISTINCT Language
FROM countrylanguage;
___

6. Aggregate Functions

SELECT COUNT(*) FROM country;  
SELECT SUM(Population) FROM country;  
SELECT AVG(Population) FROM country;  
SELECT MAX(Population) FROM country;  
SELECT MIN(Population) FROM country;  
___

7. GROUP BY — Group Data

SELECT Continent, COUNT(*) AS Number_of_Countries
FROM country
GROUP BY Continent;  
SELECT Continent, SUM(Population) AS Total_Population
FROM country
GROUP BY Continent;
___

8. HAVING — Filter Groups

SELECT Continent, COUNT(*) AS Number_of_Countries
FROM country
GROUP BY Continent
HAVING COUNT(*) > 10;
___

9. LIKE — Search for a Pattern

SELECT * FROM country
WHERE Name LIKE 'A%';  
SELECT * FROM country
WHERE Name LIKE '%stan%';
___

10. IN — Match Multiple Values

SELECT * FROM country
WHERE Continent IN ('Asia', 'Europe', 'Africa');
___

11. BETWEEN — Select a Range

SELECT Name, Population
FROM country
WHERE Population BETWEEN 10000000 AND 50000000;
___

12. IS NULL / IS NOT NULL — Check Missing Values

SELECT * FROM country
WHERE IndepYear IS NULL;  
SELECT * FROM country
WHERE IndepYear IS NOT NULL;
___

13. JOIN — Combine Tables

SELECT country.Name, city.Name
FROM country
JOIN city
ON country.Code = city.CountryCode;
___

14. LEFT JOIN

SELECT country.Name, city.Name
FROM country
LEFT JOIN city
ON country.Code = city.CountryCode;
___

15. Subquery — Query Inside Another Query

SELECT Name, Population
FROM country
WHERE Population > (  
    SELECT AVG(Population)
    FROM country
);
___
