### Importing Ordinal Encoder

<b>OrdinalEncoder</b> is used to convert **ordinal categorical data** into numerical values.

It assigns a numerical value to each category based on the **order or ranking** of the categories.

```python
from sklearn.preprocessing import OrdinalEncoder
```

___
### Defining the order of categories

Defining the **order of categories** for two ordinal categorical columns.

* For the first column, the order is:
  Poor → Average → Good
* For the second column, the order is:
  School → UG → PG

The **categories** parameter allows us to specify the correct order instead of letting the encoder decide it automatically.

The encoder will assign numerical values according to this order:

* Poor → 0, Average → 1, Good → 2
* School → 0, UG → 1, PG → 2

```python
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
x_train = oe.fit_transform(x_train)
x_test = oe.transform(x_test)
```

```python
x_train
```

___
### Checking the Categories

oe.categories_ is used to view the **categories specified by the OrdinalEncoder**.

* First column: ['Poor', 'Average', 'Good']
* Second column: ['School', 'UG', 'PG']

The **_** at the end of categories_ indicates that this is an **attribute containing information stored by the encoder** after it has been fitted.


```python
oe.categories_
```

___
### Importing LabelEncoder

LabelEncoder is used to convert categorical labels into numerical values.
It assigns a unique numerical value to each category. It is commonly used for encoding the target variable in a machine learning problem.

For example, categories such as Yes and No can be converted into numerical values such as 1 and 0.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
y_train.head()
```

```python
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
y_train
```

___
### Checking the Classes

le.classes_ displays all the categories in the order in which LabelEncoder has assigned their numerical labels.

For example, if the categories are **no,yes** the classes may be displayed in alphabetical order. The position of each category determines its encoded numerical value.


```python
le.classes_
```
