### Importing Libraries  

```python
import pandas as pd
import requests
```

___
This imports the Requests library, which allows Python to communicate with websites and APIs.So,We use it when our program needs to fetch or send data through an API.The get() function will fetch the data from the given url.A GET request is generally used when we want to retrieve/get data from a server.

Here, the server is the TMDB API.  

This URL is for Page=1 only.

```python
response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=4d4d86ef75795c17c59a31b3abcf7b4c&language=en-US&page=1')
```

___
1) Now the variable response is a object which contain many methods like(status code,headers,response content,json)  
2) We will just extract the json method which consists of the data.
3) ['results'] means:
Go inside that dictionary and give me the value associated with the key 'results'.According to your database structure, this is the Result List, containing the movie dictionaries.
4) The list of column names You're telling Pandas that From the DataFrame, I only want these seven columns and discard the other movie information returned by TMDB.

```python
udf = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

___
Creating an empty Dataframe for storing all the pages data because the previous dataframe holds information of a single page.

```python
df = pd.DataFrame()
```

___
1) The Range is from 1 to 51 which means that fetch pages from 1-50 ,although there were 428 pages but it will take so long to execute them.
2) Now the URL which is used here is not for only page 1 because we have used <b>page={}'.format(i)</b>. It replaces {} with the current value of i.
3) temp_df means temporary DataFrame.Because it contains only the movies from the current page and will be updated with the next iteration.
4) pd.concat()...This is the line that combines the newly fetched movies with the previously collected movies.Combine the existing df with the newly created temp_df.
5) When Pandas combines DataFrames, it can preserve their original row indexes.

For example:

df:0,1,2  
temp_df:0,1,2

If you concatenate them without changing the index, you could get:

0,1,2,0,1,2

There are duplicate indexes.

By using ignore_index=True

Pandas creates a new continuous index:

0,1,2,3,4,5

After combining the DataFrames, forget the old indexes and create a fresh index.

```python
for i in range(1,51): # 428 pages will so much of time so fetchinf 50 pages data.  
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={}'.format(i))
    temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    df = pd.concat([df,temp_df],ignore_index=True)
    
```
