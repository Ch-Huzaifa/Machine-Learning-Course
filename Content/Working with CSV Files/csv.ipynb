### Reading CSV File using URL:
<b>In order to read a csv file online, from a URL then we hace to :  </b>
1) First   
   import requests from io import StringIO
2) Then paste the url in the Qutations  
   url= ""
3) Place the header   
   headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
4) Then use the request module  
   req = requests.get(url, headers=headers)
5) Only for Reading Data, write :    
   data = StringIO(req.text)
6) Then read the data  
   pd.read_csv(data)

   You just have to copy this whole piece of data and use it by placinf only the url when you have to.

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### skip_rows:
If we want to skip one or more rows from the dataset than we use <b>skip_rows</b> parameter. In this case rows index will start from 1 not 0.

1) For removing the First n-rows we simply mention the value and the rows will be removed.  
Example :   
skip_rows=4  (This will remove the first 4 rows.)  

2) For removing certain rows we simply mention the rows value not index value and the rows will be removed.  
Example :   
skip_rows=[1,4]  (This will remove the first 1st and 4th row.)

3) For removing rows within a range we simply mention the range and the rows will be removed.  
Example :   
skip_rows=range(1,4)  (This will remove the rows from 1 to 3.)

### Before :

```python
pd.read_csv('aug_train.csv')
```

### After :

```python
pd.read_csv('aug_train.csv',skiprows=range(1,4))
```

## converters :
converters are used to convert the values of a column when the CSV is being loaded.

1) We have used a saperate  function <b>rename</b> to rename the full names to abbreviations in the columns <b>team1</b> and used it in the converter parameter.

2) We have also used a condition for column <b>neutral_venue</b> to increase the value by 5.

### Before :

The values in column <b>team1</b> are <b>Royal Challengers Bangalore</b> and <b>Kings XI Punjab</b> we will conert them into <b>RCB</b> and <b>K XI P</b>.

We will also increase each value in the <b>neutral-value</b> column by <b>5</b>.

```python
pd.read_csv('IPL Matches 2008-2020.csv').head()
```

### After :

1) Rename is a simple python function we use it in the conerter parameter.

2) lambda represents each value in the column and we increment each value by 5.

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    elif name == "Kings XI Punjab":
        return "K XI P"
    else:
        return name

pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename,"neutral_venue":lambda x:int(x)+5}).head()
```
